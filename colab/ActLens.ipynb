# ActLens on Colab

The model runs on the Colab GPU; you view the visualization in **your own browser** through a private link.

1. Runtime → Change runtime type → **T4 GPU**
2. Run the cell below (Runtime → Run all). The first run takes a few minutes: it installs ActLens and downloads the model.
3. Click the **Open ActLens** link it prints. Keep this Colab tab open while you use it.

The link contains a secret token: anyone with the full link can use the server, so don't share it.

```
#@title Start ActLens
MODEL = "Qwen/Qwen3-0.6B"  #@param {type:"string"}
DTYPE = "float16"  #@param ["float16", "float32"]

!pip install -q actlens

import torch
if not torch.cuda.is_available():
    print("No GPU found: switch the runtime to a GPU (Runtime → Change runtime type). Continuing on CPU, which is slow.")

from actlens.colab import launch
launch(model=MODEL, dtype=DTYPE)
```

**Troubleshooting** — if the link does not load, look at the logs:
