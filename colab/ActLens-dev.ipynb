# ActLens dev build on Colab

Smoke-test a pre-release from TestPyPI on a Colab GPU (Runtime → Change runtime type → **T4 GPU**), then run the cells in order.

Set `VERSION` to the pre-release you uploaded with `twine upload --repository testpypi dist/*`. Leave it empty to install the latest TestPyPI build.

```
#@title Install
!pip install -i https://test.pypi.org/simple/ --extra-index-url https://pypi.org/simple/ --pre actlens
!pip show actlens
!actlens --version
```

```
#@title Start ActLens
MODEL = "openai-community/gpt2"  #@param {type:"string"}
DTYPE = "float16"  #@param ["float16", "float32"]

from actlens.colab import launch
launch(model=MODEL, dtype=DTYPE)
```

**Troubleshooting** — if the link does not load, look at the logs:

```
from actlens.colab import stop
stop()
```
